```python
import pandas as pd
import os

# check path
os.getcwd()

# pull csv file from nhpi to clean 
# ../data/nhpi.csv is the file name of the csv file 
df = pd.read_csv(
    "../data/nhpi.csv",  # adjust path if notebook is in notebooks/
    sep=",",             # CSV is comma-separated
    quotechar='"'        # handle the quotes around column names
)

df.head()


# keeping REF_DATE → date
# GEO → geographic region
# New housing price indexes → component (Total / House / Land)
#VALUE → the index
# STATUS → optional quality flag

print(df.columns.tolist())

df = df[["REF_DATE", "GEO", "New housing price indexes", "VALUE", "STATUS"]]

# rename columns 
df = df.rename(columns={
    "REF_DATE": "date",
    "GEO": "geo",
    "New housing price indexes": "component",
    "VALUE": "price_index",
    "STATUS": "status"
})
# Parse Dates , get year / month 
# df["date"] = pd.to_datetime(df["date"], infer_datetime_format=True) <-- if I don't know the format 
df["date"] = pd.to_datetime(df["date"], format="%Y-%m")
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month

# Parse out missing values, and empty strings → convert to NaN:
df["price_index"] = pd.to_numeric(df["price_index"], errors="coerce")
df = df.dropna(subset=["price_index"])  # remove rows with missing index

# Clean the columns to remove extra white spaces 
# Strip whitespace
df.columns = df.columns.str.strip()
df["geo"] = df["geo"].str.strip()
df["component"] = df["component"].str.strip()

# Check unique values:
print(df["component"].unique())
print(df["geo"].unique())

# Remove duplicates 
df = df.drop_duplicates()

# Check rows after cleaning 
print("Rows after cleaning:", len(df))

# Save CSV file 
df.to_csv("../data/nhpi_cleaned.csv", index=False)






```
